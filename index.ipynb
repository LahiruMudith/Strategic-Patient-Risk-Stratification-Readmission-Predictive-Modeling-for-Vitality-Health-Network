* **`.info()`**

* **`.describe()`**
    Generates descriptive statistics that summarize the central tendency of the data. Key statistics for object (string) data include:
    * **count:** Total number of non-null values.
    * **unique:** Number of distinct values.
    * **top:** The most common value (the mode).
    * **freq:** The frequency (occurrence count) of the top value.

* **`.head()`**
    Returns the first *n* rows of the DataFrame. By default, `n=5`.

* **`.columns`**
    Print the column names of the DataFrame.
